```python
import json
from pathlib import Path
import matplotlib.pyplot as plt

Path("results/graphs").mkdir(parents=True, exist_ok=True)

for name in ["mlp", "cnn"]:
    with open(f"results/{name}_evaluation.json", "r") as file:
        result = json.load(file)

    history = result["history"]

    # Training and validation accuracy
    plt.figure(figsize=(7, 4))
    plt.plot(history["accuracy"], label="Training accuracy")
    plt.plot(history["val_accuracy"], label="Validation accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.title(f"{name.upper()} - Accuracy")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"results/graphs/{name}_accuracy.png", dpi=150)
    plt.show()
    plt.close()

    # Training and validation loss
    plt.figure(figsize=(7, 4))
    plt.plot(history["loss"], label="Training loss")
    plt.plot(history["val_loss"], label="Validation loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title(f"{name.upper()} - Loss")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"results/graphs/{name}_loss.png", dpi=150)
    plt.show()
    plt.close()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

Path("results/graphs").mkdir(parents=True, exist_ok=True)

for name in ["mlp", "cnn"]:
    matrix = np.loadtxt(
        f"results/{name}_confusion_matrix.csv",
        delimiter=",",
        dtype=int
    )

    plt.figure(figsize=(8, 6))
    plt.imshow(matrix, interpolation="nearest")
    plt.title(f"{name.upper()} - Confusion Matrix")
    plt.xlabel("Predicted digit")
    plt.ylabel("Actual digit")
    plt.xticks(range(10))
    plt.yticks(range(10))
    plt.colorbar()

    for row in range(10):
        for col in range(10):
            plt.text(col, row, str(matrix[row, col]),
                     ha="center", va="center", fontsize=8)

    plt.tight_layout()
    plt.savefig(f"results/graphs/{name}_confusion_matrix.png", dpi=150)
    plt.show()
    plt.close()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from pathlib import Path

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_test_normalised = x_test.astype("float32")[..., None] / 255.0

Path("results/misclassified").mkdir(parents=True, exist_ok=True)

for name in ["mlp", "cnn"]:
    model = tf.keras.models.load_model(f"models/{name}.keras")

    probabilities = model.predict(x_test_normalised, verbose=0)
    predictions = probabilities.argmax(axis=1)
    wrong_indices = np.where(predictions != y_test)[0]

    print(name.upper(), "misclassified images:", len(wrong_indices))

    selected = wrong_indices[:12]

    fig, axes = plt.subplots(3, 4, figsize=(10, 7))

    for ax, index in zip(axes.flat, selected):
        ax.imshow(x_test[index], cmap="gray")
        ax.set_title(
            f"Actual: {y_test[index]}, Predicted: {predictions[index]}"
        )
        ax.axis("off")

    for ax in axes.flat[len(selected):]:
        ax.axis("off")

    plt.tight_layout()
    plt.savefig(f"results/misclassified/{name}_examples.png", dpi=150)
    plt.show()
    plt.close()
```
